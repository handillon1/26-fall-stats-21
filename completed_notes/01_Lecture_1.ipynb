# AI: Artificial Intelligence and Academic Integrity

## Why do the work yourself?

- Imagine an AI tool can produce working solutions to every assignment in this class.
- Having a finished solution does not tell you whether **you** understand it or could solve a similar problem.
- The assignments give you practice reasoning about data, reading code, and checking results.
- **The goal of this class is to develop understanding and skills you can use yourself.**


## No Pain, No Gain

Think of the gym. Lifting weights is a means to the real goal of gaining strength. The benefit comes from doing the training yourself.

Learning to code also takes practice: breaking a problem into smaller steps, trying an approach, and working through mistakes.

When you get stuck, use that moment to ask: What do I expect to happen? What actually happened? What could explain the difference?

**The effort you put into understanding the problem is part of the learning.**


## No Pain, No Gain (Continued)

Having AI solve a difficult assignment (or copying a solution) is like having a stronger person lift the weights for you.

The weights have moved, but you have missed the training.

In the same way, receiving a finished solution can skip the practice that the assignment is designed to give you.

**Practice now builds the understanding you will need to direct and evaluate computational work later.**


## Course Goals

It is easy to confuse completing coursework with learning from it.

When getting a finished answer becomes the priority, it is tempting to skip the practice the assignment was intended to provide.

**The goal of the course is your learning.**

I will admit, a major conflicting issue here is that I am not able to create individualized grading schemas that evaluate exactly how much each student learned over the course. All students are graded on the same criteria and evaluated on what they turn in for the assignments.

That said, I hope you can judge your performance in a class based on what you learned and not your letter grade.


## My expectations

When you face a challenging homework assignment, I expect:
- you work hard
- you will not seek out solutions from AI or another student or some other resource 
- if you are not able to complete everything required by the assignment by the deadline
    + you submit what you have and accept a grade that is less than 100%
    + you view this not as a failure of your coding abilities, but as a indication of areas for growth and improvement

*I (and the statistics department) take issues of plagiarism seriously and will escalate cases to the Dean of Students. Full details regarding academic integrity are in the syllabus.*

## AI Collaboration Policy

Read the course policies on AI and collaboration and be sure you understand it.

**You are encouraged to discuss code that is not part of an assignment!**

This is a coding class! As long as the code is not part of a homework assignment, you can post and discuss code with each other and also on Campuswire.

## Why learn programming when AI can write code?

Even if AI writes the code, you still need to understand the problem, decide what a good result looks like, and evaluate the answer.

- Programming knowledge gives you a vocabulary for describing what you want.
- Experience helps you ask useful questions, spot questionable assumptions, and investigate unexpected results.
- Understanding the work helps you explain and take responsibility for the conclusions.

**These skills remain useful as the tools become more capable.**


## Human knowledge gives code its purpose

- **Purpose:** What question are we answering, and who will use the answer?
- **Context:** What does each row represent, and how were the data collected?
- **Assumptions:** Does a zero mean a measured value of zero, or a missing observation?
- **Interpretation:** Does the result answer the original question? What conclusions does it support?

**A program can run successfully while computing the wrong thing.**


## The skills we will practice

- **Specify:** Describe a function's inputs and expected output: value, type, shape, and units.
- **Predict:** Work through a small example by hand so you know what should happen.
- **Check and debug:** Compare the output with your prediction, investigate differences, and try edge cases.
- **Explain:** Describe the code, results, and assumptions in your own words.

**Example:** A function averaging `[80, 90, 100]` should return one number, `90`. What should it do with an empty list?


# Grades and Life

## Your grades do not define you

You are here at UCLA. One of the reasons you got into UCLA was because you had good grades in high school and/or at community college. While you are in school, a lot of your energy is poured into your classes and I can understand why grades feel so important. 

That said, **Your grades do not define you.**

It feels good to get good grades. Grades do play a role in graduate school admissions. But they are not the most important thing in life. No one on their death bed looks back and says "I wish I got an A- instead of a B+ in that one college class."

## Work - Life Balance

I like to split where we put our energies of life into three broad categories:

1. **Work**
    + Jobs and internships
    + School and academics
    + Other professional obligations
2. **Relationships**
    + Family
    + Friends
    + Romantic partner
    + Other social obligations
3. **Self**
    + Care of physical health (food, sleep, exercise)
    + Care of mental health (sleep, play, entertainment)
    + Care of spiritual health (if you are spiritual/religious)

*There are 24 hours in a day. It is not possible to give 100% to all categories.*

## Work - Life Balance (Trade-offs)

Work-Life balance is achieved by consciously choosing what is important to you and devoting your time and energies accordingly.

In general, the more you put in, the more you get out.

Satisfaction can be found by accepting the natural consequences of what you have chosen to deprioritize.

**Example:** Investing all relationship hours into a romantic partner means naturally growing distant from friends. Accepting the natural consequences of investing less time into something can reduce feelings of bitterness and jealousy.

## Work - Life Balance (Corporate World)

In the corporate and professional world, people who devote a lot of energy into the goals of the company are rewarded. The company is not necessarily punishing people who choose to have families and a life outside of work.

From the company's perspective, who would they rather promote?
+ the person who did everything asked of them and then continued to stay at work and did even more
+ the person who did everything asked of them and then immediately left to spend time with their family/friends/romantic partner

You have to choose what is important to you and spend your time accordingly.

## Self care is important

You must not neglect taking care of your physical and mental health.

If you neglect care of self, you will likely operate at less than 100% efficiency and the time you invest in work/school/relationships will not be as productive.

**Examples:**
- Choosing to hang out with a friend instead of sleeping, resulting in being 'out of it'. (Maybe decline and get sleep!)
- Skipping meals/sleep to study for exams, resulting in getting sick and poor performance. (Maybe eat properly, sleep well, and study a bit less!)

## Self care is important (Continued)

When I tell you that your physical and mental health is important, I'm encouraging you to choose to invest your time into activities like exercise, sleep, and relaxation.

Sometimes this means choosing not to complete your homework to 100%. The natural consequence of this is a homework grade that is less than 100%.

When you can readily embrace this natural consequence of prioritizing your own physical and mental health over your homework grade, you can enjoy the quarter with less bitterness, more joy and better health.

## Beware of "fruitless" entertainment

Entertainment and fun activities are important for your mental well being.

Participating in an entertainment activity should be a break from work and should give you mental energy so you can return to your work in a good mood.

Some activities can have the opposite effect - they drain you. Some video games, apps, and social media sites are designed to be addictive. After spending hours doing the activity, you don't feel good about yourself.

**Be mindful and selective about your entertainment activities.**

# Shell Basics

## Opening Terminal on MacOS

- **Probably the quickest:** Open spotlight with `Command + space`. Start typing "Terminal". Terminal will appear as the top hit after you type the first few letters. Hit Enter to start.
- **Another method:** You can open Launchpad from the dock. Click "Other". Click "Terminal".

*I suggest adding the terminal to the dock.*

## Getting Help

**Windows PowerShell**
```bash
help commandname
```

**Mac OS and git bash**
```bash
commandname --help
```

Type `q` to exit help. Hit the space bar to scroll to the next page.

## Shell Basics: Navigation

- `~` is your home directory.
- `pwd` will tell you where you are currently located. (present working directory)
- `cd` is the command to change your directory

Wherever you are, you can switch to your home directory with `cd ~`

## Shell Basics: Navigation (Hierarchy)

Directories are listed in a hierarchy. For example, `home/classes/26-fall-stats-21`

- `cd homework` will change to the directory `homework` if it is available in your current directory.
- `cd ..` will take you to the parent directory.
- `cd ../..` will take you two levels up.

**Tab Completion:** Shell supports tab completion. Begin by typing `cd ho` and then hit `TAB`. Shell will try to complete what you are typing.

## Shell Basics: Managing Directories

- `ls` will list the contents of your current directory.
- `mkdir name` will create a new directory called `name` inside your current working directory.
- `clear` will clear the screen.
